```python
# PROJECT: Employee Retention and Pattern Analysis
'''
Problem Overview:
We are working on a people-analytics task for a technology company. In this project, we will examine employee records covering attendance, work tenure, job satisfaction, overtime, and attrition. The main objective is to identify group-level trends and patterns that the HR team can look into further—rather than judging or predicting the behavior of any individual employee.

Questions to Investigate:
1. What is the total number of employees in the dataset, and how is the workforce divided across different departments?
2. What is the overall employee attrition (leaving) rate across the company?
3. Which specific departments or job roles are showing higher employee turnover?
4. How does the attrition pattern change across different employee tenure or experience groups?
5. Is there a noticeable difference in job satisfaction levels between employees who left and those who stayed?
6. How do overtime hours and attendance percentages compare between employees who stayed and those who resigned?
7. What extra information or data would the HR team need before taking final action?
8. Which practical retention initiatives can be tested and evaluated to reduce turnover?
'''


import numpy as np
import pandas as pd

# Step 1: Create Sample Dataset
rng = np.random.default_rng(42)
total_rows = 120

departments = np.array(["IT"]*36 + ["Sales"]*30 + ["Support"]*24 + ["HR"]*18 + ["Finance"]*12)
rng.shuffle(departments)

roles = rng.choice(["Junior", "Mid-Level", "Senior"], size=total_rows, p=[0.5, 0.3, 0.2])
tenure = rng.integers(1, 11, size=total_rows)
satisfaction = rng.integers(1, 5, size=total_rows)
overtime = rng.choice(["Yes", "No"], size=total_rows, p=[0.35, 0.65])
attendance = np.round(rng.uniform(72.0, 99.0, size=total_rows), 1)

attrition_list = []
for i in range(total_rows):
    score = 0
    if overtime[i] == "Yes": score += 1
    if satisfaction[i] <= 2: score += 1
    if attendance[i] < 82.0: score += 1
    if tenure[i] <= 2: score += 1
    attrition_list.append("Yes" if score >= 2 else "No")

df = pd.DataFrame({
    "Employee_ID": [f"EMP{1001 + i}" for i in range(total_rows)],
    "Department": departments,
    "Role": roles,
    "Tenure_Years": tenure,
    "Job_Satisfaction": satisfaction,
    "Overtime": overtime,
    "Attendance_Pct": attendance,
    "Attrition": attrition_list
})
df["Left_Flag"] = np.where(df["Attrition"] == "Yes", 1, 0)

# Q1: Total Employees & Department Distribution (Set, Dict, Loop)
total_emp = len(df)
unique_depts = set(df["Department"])
dept_counts = {}
for d in df["Department"]:
    if d in dept_counts:
        dept_counts[d] += 1
    else:
        dept_counts[d] = 1

print("--- Q1: WORKFORCE DISTRIBUTION ---")
print("Total Employees:", total_emp)
print("Departments (Set):", unique_depts)
print("Department Counts (Dict):", dept_counts)

# Q2: Overall Attrition Rate
left_emp = len(df[df["Attrition"] == "Yes"])
attrition_rate = round((left_emp / total_emp) * 100, 2)

print("\n--- Q2: OVERALL ATTRITION RATE ---")
print(f"Employees Left: {left_emp} out of {total_emp} ({attrition_rate}%)")

# Q3: Attrition by Department & Job Role
print("\n--- Q3: ATTRITION BY DEPARTMENT (%) ---")
print((df.groupby("Department")["Left_Flag"].mean() * 100).round(2))
print("\n--- Q3: ATTRITION BY ROLE (%) ---")
print((df.groupby("Role")["Left_Flag"].mean() * 100).round(2))

# Q4: Attrition by Tenure Group (Function & If-Elif-Else)
def get_tenure_group(years):
    if years <= 2:
        return "0-2 Years"
    elif years <= 5:
        return "3-5 Years"
    else:
        return "6+ Years"

df["Experience_Group"] = df["Tenure_Years"].apply(get_tenure_group)
print("\n--- Q4: ATTRITION BY TENURE GROUP (%) ---")
print((df.groupby("Experience_Group")["Left_Flag"].mean() * 100).round(2))

# Q5: Job Satisfaction Comparison (Tuple)
rating_scale = (1, 2, 3, 4)
print("\n--- Q5: JOB SATISFACTION (NO = STAYED, YES = LEFT) ---")
print("Rating Scale (Tuple):", rating_scale)
print(df.groupby("Attrition")["Job_Satisfaction"].mean().round(2))

# Q6: Overtime & Attendance Comparison (NumPy & Crosstab)
stayed_att = round(np.mean(df[df["Attrition"] == "No"]["Attendance_Pct"]), 2)
left_att = round(np.mean(df[df["Attrition"] == "Yes"]["Attendance_Pct"]), 2)

print("\n--- Q6: OVERTIME & ATTENDANCE COMPARISON ---")
print(f"Avg Attendance (Stayed): {stayed_att}% | Avg Attendance (Left): {left_att}%")
print("\nOvertime vs Attrition Count:")
print(pd.crosstab(df["Overtime"], df["Attrition"]))

```
